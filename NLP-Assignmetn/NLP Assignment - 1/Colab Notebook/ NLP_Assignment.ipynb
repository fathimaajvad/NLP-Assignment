```python
#Import Libraries
import nltk
import string

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
#Convert to Lowercase
text = """My name is Fathima P Ajvad.
I am from Aluva, Kerala.
I am currently pursuing my bachelors degree in computer science and engineering with specialization in artificial intelligence at Adi Shankara Institute of Engineering and Technology Kalady.
I completed my 12th from GHSS Pazhayannur.
My family consists of 6 members.
I have done an internship with Mulearn Foundation and also participated in various hackathons.
I have also won a prize at Build on Rails Hackahton conducted as part of IEDC Summit.
My hobbies are travelling and baking.
I love to explore the field of AI and ML in deep and I read latest publications by DeepMind to stay in the loop.
My strengths are I can adapt well and I can do a bit of multitasking.
My weaknesses are I find it difficult to delegate work to people as I try to be perfectionist eveytime and also I fail in meeting the deadlines sometimes.
I am actively trying to convert my weaknesses to my strenghs."""
lower_text = text.lower()
print("Lowercase Text:\n")
print(lower_text)
```

```python
#Sentence Tokenization
import nltk
nltk.download('punkt_tab')
sentences = sent_tokenize(lower_text)

print("Sentence Tokenization:\n")
for sentence in sentences:
    print(sentence)
```

```python
#Word Tokenization
words = word_tokenize(lower_text)

print(words)
```

```python
#Remove Punctuation
words_no_punct = [word for word in words if word not in string.punctuation]

print(words_no_punct)
```

```python
#Remove Stopwords
stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words_no_punct if word not in stop_words]

print(filtered_words)
```

```python
#Stemming
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]

print(stemmed_words)
```

```python
#Lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print(lemmatized_words)
```

```python
#Final Preprocessed Text
final_text = " ".join(lemmatized_words)

print("Final Preprocessed Text:\n")
print(final_text)
```
